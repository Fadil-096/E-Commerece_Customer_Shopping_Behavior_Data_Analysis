```python
import pandas as pd
data = pd.read_csv('data/raw/E-Commerce_Data.csv')
```

```python
#Handling the missing values (Filling with the meadian values as it is not senstitive to outliers)
data['Review Rating'] = data.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))  
```

```python
data.isnull().sum()
```

```python
data.columns
```

```python
data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(' ','_')
```

```python
data = data.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
data.columns
```

```python
labels = ['Young Adults','Adults','Middle-aged','Senior']
data['age_group'] = pd.qcut(data['age'], q=4, labels = labels)
```

```python
data[['age', 'age_group']].head()
```

```python
data.head()
```

```python
frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

data['purchase_frequency_days'] = data['frequency_of_purchases'].map(frequency_mapping)
```

```python
data[['purchase_frequency_days', 'frequency_of_purchases']].head()
```

```python
data[['discount_applied','promo_code_used']].head()
```

```python
(data['discount_applied'] == data['promo_code_used']).all()
```

```python
#Removing one column as discount_applied column already represents the same
data = data.drop('promo_code_used', axis = 1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
numeric_cols = ['age', 'purchase_amount', 'review_rating', 'previous_purchases']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.flatten(), numeric_cols):
    sns.histplot(data[col], kde=True, ax=ax)
    ax.set_title(f'Distribution of {col}')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, len(numeric_cols), figsize=(14, 5))
for ax, col in zip(axes, numeric_cols):
    sns.boxplot(y=data[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
corr = data[numeric_cols + ['purchase_frequency_days']].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()

username = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")
host = os.getenv("DB_HOST")
port = os.getenv("DB_PORT")
database = os.getenv("DB_NAME")

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

table_name = "customer"
data.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
